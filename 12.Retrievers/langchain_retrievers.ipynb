# **Wikipedia Retriever**

**Wikipedia Retriever**

A retriever is a component that takes a user query and returns relevant documents.

The Wikipedia Retriever searches Wikipedia using the Wikipedia API and returns the relevant Wikipedia articles as LangChain `Document` objects.

Unlike a vector store retriever, Wikipedia Retriever does not require an embedding model or Chroma database.

**1. Install required packages**



```python
!pip install -U langchain-community wikipedia
```

**2. Import WikipediaRetriever**

`WikipediaRetriever` is the LangChain retriever used to search Wikipedia and return relevant documents.

```python
from langchain_community.retrievers import WikipediaRetriever
```

**3. Initialize the Wikipedia Retriever(Create the Retriever)**

`top_k_results` controls how many Wikipedia results are returned.

`lang="en"` specifies that we want to search the English Wikipedia.

```python
retriever = WikipediaRetriever(
    top_k_results=2,
    lang="en"
)
```

**4. Define the Query**

The query is the question or topic that we want to search for on Wikipedia.

```python
query = "What is AI"
```

**5. Retrieve Relevant Documents**

`invoke()` sends the query to the retriever.

The retriever searches Wikipedia and returns a list of relevant LangChain `Document` objects.

```python
docs = retriever.invoke(query)
```

**6. View Retrieved Documents**

The returned value is a list of LangChain `Document` objects.

Each document can contain:

- `page_content` → the retrieved Wikipedia text
- `metadata` → information about the source/article

```python
docs
```

**8. Print the Retrieved Content**

```python
for i, doc in enumerate(docs):
    print(f"\n--- Result {i+1} ---")
    print(f"Content:\n{doc.page_content}")
    print(f"\nMetadata:\n{doc.metadata}")
```

# **Vector Store Retriever**

**Vector Store Retriever**

A Vector Store Retriever retrieves relevant documents from a vector store based on the similarity between the user's query and the stored document embeddings.

In this example:
- Gemini is used as the embedding model.
- Chroma is used as the vector store.
- The vector store is converted into a retriever using `as_retriever()`.

**1. Install Required Packages**

This cell installs the packages required for Gemini embeddings and the Chroma vector store.

```python
!pip install -U langchain-google-genai langchain-chroma chromadb
```

**2. Import Required Classes**

`GoogleGenerativeAIEmbeddings` converts text into numerical vectors.

`Chroma` stores those vectors and allows similarity-based retrieval.

`Document` represents a LangChain document containing text and optional metadata.

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

**3. Create Source Documents**

Each `Document` contains text that will be converted into an embedding and stored in Chroma.

```python
# Step 1: Your source documents
documents = [
    Document(page_content="LangChain helps developers build LLM applications easily."),
    Document(page_content="Chroma is a vector database optimized for LLM-based search."),
    Document(page_content="Embeddings convert text into high-dimensional vectors."),
    Document(page_content="OpenAI provides powerful embedding models."),
]
```

**4. Initialize Gemini Embedding Model**

The embedding model converts each document into a vector representation.

We use `gemini-embedding-001` and 32 dimensions for this learning example.

```python
from google.colab import userdata
import os

from langchain_google_genai import GoogleGenerativeAIEmbeddings

# Load Gemini API key from Colab Secrets
os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")

# step 2: Create Gemini embedding model
embedding_model = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    output_dimensionality=32
)
```

**5. Create Chroma Vector Store**

`Chroma.from_documents()` creates a Chroma vector store from the documents and their embeddings.

Because `persist_directory` is not specified here, this example uses an in-memory/ephemeral collection.

```python
# Step 3: Create Chroma vector store in memory
vectorstore = Chroma.from_documents(
    documents=documents,
    embedding=embedding_model,
    collection_name="my_collection"
)
```

**6. Convert Vector Store into a Retriever**

`as_retriever()` converts the Chroma vector store into a LangChain Retriever.

`k=2` means the retriever should return the 2 most relevant documents.

After this conversion, we can use `retriever.invoke(query)` to retrieve relevant documents.

```python
# Step 4: Convert vectorstore into a retriever
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 2}
)
```

**7. Define the Query**

The query is the question or topic for which we want to retrieve relevant documents.

```python
query = "What is Chroma used for?"
```

**8. Retrieve Relevant Documents**

`invoke()` sends the query to the retriever.

The retriever performs similarity-based retrieval and returns the most relevant documents.

```python
results = retriever.invoke(query)
```

**9. Display Retrieved Documents**

The retriever returns a list of LangChain `Document` objects.

Each document contains its `page_content` and any associated metadata.

```python
for i, doc in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content)
```

# **MMR Retriever**

**1. Set Up Gemini API Key**

This cell loads the Gemini API key securely from Google Colab Secrets.

The API key is required because Gemini is being used as the embedding model.

```python
from google.colab import userdata
import os

os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")
```

**2. Install Required Packages**

This cell installs the packages required for Gemini embeddings, FAISS, and LangChain community integrations.

```python
!pip install -U langchain-google-genai langchain-community faiss-cpu
```

**3. Create Sample Documents**

This cell creates sample LangChain documents that will be converted into embeddings and stored in the FAISS vector store.

```python
from langchain_core.documents import Document
# Sample documents
docs = [
    Document(page_content="LangChain makes it easy to work with LLMs."),
    Document(page_content="LangChain is used to build LLM based applications."),
    Document(page_content="Chroma is used to store and search document embeddings."),
    Document(page_content="Embeddings are vector representations of text."),
    Document(page_content="MMR helps you get diverse results when doing similarity search."),
    Document(page_content="LangChain supports Chroma, FAISS, Pinecone, and more."),
]
```

**4. Create FAISS Vector Store**

This cell creates a FAISS vector store using the Gemini embedding model.

FAISS stores the document embeddings and allows similarity-based retrieval.

```python
from langchain_community.vectorstores import FAISS
from langchain_google_genai import GoogleGenerativeAIEmbeddings
# Initialize embedding model
embedding_model = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    output_dimensionality=32
)
# Step 2: Create the FAISS vector store from documents
vectorstore = FAISS.from_documents(
    documents=docs,
    embedding=embedding_model
)
```

**5. Create an MMR Retriever**

This cell converts the FAISS vector store into a retriever and enables MMR (Maximal Marginal Relevance).

MMR tries to select documents that are both relevant to the query and sufficiently different from each other.

`k=3` means that 3 documents will be returned.

`lambda_mult=0.5` balances relevance and diversity.

```python
# Enable MMR in the retriever
retriever = vectorstore.as_retriever(
    search_type="mmr",        # <-- This enables MMR
    search_kwargs={
        "k": 3,               # k = top results
        "lambda_mult": 0.5    # lambda_mult = relevance-diversity balance
    }
)
```

**6. Define the Query**

This cell defines the question that will be sent to the MMR retriever.

```python
query = "What is LangChain?"
```

**7. Retrieve Documents Using MMR**

This cell sends the query to the retriever.

The MMR retriever selects documents that are relevant to the query while also trying to avoid returning documents that contain very similar information.

```python
results = retriever.invoke(query)
```

**8. Display Retrieved Documents**

This cell prints the documents returned by the MMR retriever.

```python
for i, doc in enumerate(results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content)
```

# **MultiQuery Retriever**

**1. Set Up Gemini API Key**

This cell loads the Gemini API key securely from Google Colab Secrets.

The API key is required because Gemini will be used as the language model and embedding model.

```python
from google.colab import userdata
import os

os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")
```

**2. Install Required Packages**

This cell installs the packages required for Gemini, FAISS, LangChain community integrations, and the MultiQuery Retriever.

```python
!pip install -U langchain-google-genai langchain-community langchain-classic faiss-cpu
```

**3. Import Required Classes**

This cell imports the classes required to create documents, generate Gemini embeddings, create the FAISS vector store, use Gemini as the LLM, and create the MultiQuery Retriever.

```python
from langchain_core.documents import Document

from langchain_community.vectorstores import FAISS

from langchain_google_genai import (
    GoogleGenerativeAIEmbeddings,
    ChatGoogleGenerativeAI
)

from langchain_classic.retrievers.multi_query import MultiQueryRetriever
```

**4. Create Sample Documents**

This cell creates sample health, technology, sports, and science documents.

These documents will be converted into embeddings and stored in FAISS.

```python
# Relevant health & wellness documents
all_docs = [
    Document(page_content="Regular walking boosts heart health and can reduce symptoms of depression.", metadata={"source": "H1"}),
    Document(page_content="Consuming leafy greens and fruits helps detox the body and improve longevity.", metadata={"source": "H2"}),
    Document(page_content="Deep sleep is crucial for cellular repair and emotional regulation.", metadata={"source": "H3"}),
    Document(page_content="Mindfulness and controlled breathing lower cortisol and improve mental clarity.", metadata={"source": "H4"}),
    Document(page_content="Drinking sufficient water throughout the day helps maintain metabolism and energy.", metadata={"source": "H5"}),
    Document(page_content="The solar energy system in modern homes helps balance electricity demand.", metadata={"source": "I1"}),
    Document(page_content="Python balances readability with power, making it a popular system design language.", metadata={"source": "I2"}),
    Document(page_content="Photosynthesis enables plants to produce energy by converting sunlight.", metadata={"source": "I3"}),
    Document(page_content="The 2022 FIFA World Cup was held in Qatar and drew global energy and excitement.", metadata={"source": "I4"}),
    Document(page_content="Black holes bend spacetime and store immense gravitational energy.", metadata={"source": "I5"}),
]
```

**5. Create Gemini Embeddings and FAISS Vector Store**

This cell creates the Gemini embedding model and uses it to convert the documents into vectors.

FAISS stores the vectors and allows similarity-based retrieval.

```python
# Initialize embedding Model
embedding_model = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    output_dimensionality=32
)
# Create FAISS vector store
vectorstore = FAISS.from_documents(
    documents=all_docs,
    embedding=embedding_model
)
```

**6. Create Similarity Retriever and MultiQuery Retriever**

The similarity retriever searches the vector store directly using the original query.

The MultiQuery Retriever uses an LLM to generate multiple variations of the user's query. It then retrieves documents for each generated query and combines the unique results.

Here, Gemini is used as the language model for generating the alternative queries.

```python
# Create retrievers
similarity_retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 5}
)

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite"
)

multiquery_retriever = MultiQueryRetriever.from_llm(
    retriever=vectorstore.as_retriever(
        search_kwargs={"k": 5}
    ),
    llm=llm
)
```

**7. Define the Query**

This cell defines the original question that will be used with both the normal similarity retriever and the MultiQuery Retriever.

```python
# Query
query = "How to improve energy levels and maintain balance?"
```

**8. Retrieve Results**

This cell sends the query to both retrievers.

The similarity retriever uses the original query directly.

The MultiQuery Retriever first generates multiple versions of the query and performs retrieval using those different queries.

```python
# Retrieve results
similarity_results = similarity_retriever.invoke(query)
multiquery_results = multiquery_retriever.invoke(query)
```

**9. Display Retrieved Results**

This cell displays the documents returned by the normal similarity retriever and the MultiQuery Retriever so that their results can be compared.

```python
print("========== SIMILARITY SEARCH ==========")

for i, doc in enumerate(similarity_results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content)


print("\n" + "*" * 100)


print("========== MULTIQUERY RETRIEVER ==========")

for i, doc in enumerate(multiquery_results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content)
```

# **Contextual Compression Retriever**

```python
from google.colab import userdata
import os

os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")
```

**1. Install Required Packages**

This cell installs the packages required for Gemini, FAISS, LangChain community integrations, and the contextual compression retriever.

```python
!pip install -U langchain-google-genai langchain-community langchain-classic faiss-cpu
```

**2. Import Required Classes**

This cell imports the classes required to create documents, generate Gemini embeddings, create a FAISS vector store, use Gemini as the language model, and create the Contextual Compression Retriever.

```python
from langchain_core.documents import Document

from langchain_community.vectorstores import FAISS

from langchain_google_genai import (
    GoogleGenerativeAIEmbeddings,
    ChatGoogleGenerativeAI
)

from langchain_classic.retrievers import ContextualCompressionRetriever
from langchain_classic.retrievers.document_compressors import LLMChainExtractor
```

**3. Create Sample Documents**

This cell creates sample documents containing information about different topics.

Some documents contain information that is relevant to photosynthesis, while other parts contain unrelated information.

This helps demonstrate how contextual compression removes content that is not relevant to the query.

```python
docs = [
    Document(
        page_content=(
            """The Grand Canyon is one of the most visited natural wonders in the world.
            Photosynthesis is the process by which green plants convert sunlight into energy.
            Millions of tourists travel to see it every year. The rocks date back millions of years."""
        ),
        metadata={"source": "Doc1"}
    ),

    Document(
        page_content=(
            """In medieval Europe, castles were built primarily for defense.
            The chlorophyll in plant cells captures sunlight during photosynthesis.
            Knights wore armor made of metal. Siege weapons were often used to breach castle walls."""
        ),
        metadata={"source": "Doc2"}
    ),

    Document(
        page_content=(
            """Basketball was invented by Dr. James Naismith in the late 19th century.
            It was originally played with a soccer ball and peach baskets. NBA is now a global league."""
        ),
        metadata={"source": "Doc3"}
    ),

    Document(
        page_content=(
            """The history of cinema began in the late 1800s. Silent films were the earliest form.
            Thomas Edison was among the pioneers. Photosynthesis does not occur in animal cells.
            Modern filmmaking involves complex CGI and sound design."""
        ),
        metadata={"source": "Doc4"}
    )
]
```

**4. Create Gemini Embeddings and FAISS Vector Store**

This cell creates the Gemini embedding model and uses it to convert the documents into vectors.

FAISS stores these vectors and allows similarity-based retrieval.

```python
embedding_model = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    output_dimensionality=32
)
# Create a FAISS vector store from the documents
vectorstore = FAISS.from_documents(
    documents=docs,
    embedding=embedding_model
)
```

**5. Create the Base Retriever**

This cell converts the FAISS vector store into a retriever.

The base retriever first finds documents that are relevant to the query.

Here, `k=5` means that up to 5 relevant documents can be retrieved before compression.

```python
base_retriever = vectorstore.as_retriever(
    search_kwargs={"k": 5}
)
```

**6. Create the Document Compressor**

This cell creates an LLM-based compressor.

`LLMChainExtractor` uses the language model to extract only the parts of each retrieved document that are relevant to the user's query.

Gemini is used as the language model.

```python
# Set up the compressor using an LLM
llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite"
)

compressor = LLMChainExtractor.from_llm(llm)
```

**7. Create the Contextual Compression Retriever**

This cell combines the base retriever and the compressor.

First, the base retriever retrieves relevant documents.

Then, the compressor removes or extracts information that is not relevant to the query.

The final result contains the most useful parts of the retrieved documents.

```python
# Create the contextual compression retriever
compression_retriever = ContextualCompressionRetriever(
    base_retriever=base_retriever,
    base_compressor=compressor
)
```

**8. Define the Query**

This cell defines the question that will be used to retrieve and compress the documents.

```python
query = "What is photosynthesis?"
```

**9. Retrieve Compressed Results**

This cell sends the query to the Contextual Compression Retriever.

The retriever first finds relevant documents and then uses the Gemini language model to keep only the parts that are relevant to the query.

```python
compressed_results = compression_retriever.invoke(query)
```

**10. Display Compressed Results**

This cell prints the compressed documents returned by the contextual compression retriever.

The output should contain information related to photosynthesis while removing unrelated information from the original documents.

```python
for i, doc in enumerate(compressed_results):
    print(f"\n--- Result {i+1} ---")
    print(doc.page_content)
```
