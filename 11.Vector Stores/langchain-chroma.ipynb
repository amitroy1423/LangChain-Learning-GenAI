**`Cell 3: Import Required Classes`**

GoogleGenerativeAIEmbeddings converts text into numerical vectors using Gemini embeddings. Chroma is used to store and search those vectors.

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_chroma import Chroma
```

**`Cell 4: Create Documents`**

This cell creates five LangChain Document objects containing information about IPL players.
Each document has:



*   page_content → the actual text
*   metadata → additional information such as the player's team



LangChain's Document class is designed to store text together with metadata for retrieval and vector-store workflows.

```python
from langchain_core.documents import Document

# Create LangChain documents for IPL players

doc1 = Document(
    page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL history. Known for his aggressive batting style and fitness, he has led the Royal Challengers Bangalore in multiple seasons.",
    metadata={"team": "Royal Challengers Bangalore"}
)

doc2 = Document(
    page_content="Rohit Sharma is the most successful captain in IPL history, leading Mumbai Indians to five titles. He's known for his calm demeanor and ability to play big innings under pressure.",
    metadata={"team": "Mumbai Indians"}
)

doc3 = Document(
    page_content="MS Dhoni, famously known as Captain Cool, has led Chennai Super Kings to multiple IPL titles. His finishing skills, wicketkeeping, and leadership are legendary.",
    metadata={"team": "Chennai Super Kings"}
)

doc4 = Document(
    page_content="Jasprit Bumrah is considered one of the best fast bowlers in T20 cricket. Playing for Mumbai Indians, he is known for his yorkers and death-over expertise.",
    metadata={"team": "Mumbai Indians"}
)

doc5 = Document(
    page_content="Ravindra Jadeja is a dynamic all-rounder who contributes with both bat and ball. Representing Chennai Super Kings, his quick fielding and match-winning performances make him a key player.",
    metadata={"team": "Chennai Super Kings"}
)
```

**`Cell 5: Store Documents in a List`**

This cell combines the five individual Document objects into one Python list. This list will later be inserted into the vector store.

```python
docs = [doc1, doc2, doc3, doc4, doc5]

print(len(docs))
```

**`Cell 6: Create Embedding Model and Vector Store`**
This cell initializes the Gemini embedding model and creates a Chroma vector store. The Gemini model will convert the document text into vectors when the documents are added.

```python
embedding = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    output_dimensionality=32
)

vector_store = Chroma(
    embedding_function=embedding,
    persist_directory="my_chroma_db",
    collection_name="sample"
)
```

**`Cell 7: Add Documents to Chroma`**
This cell sends the five documents to the vector store. Their text is converted into embedding vectors and stored in Chroma together with their metadata.

```python
# Add documents to the vector store

vector_store.add_documents(docs)

print("Documents added successfully!")
```

**`Cell 8: Inspect the Vector Store`**
This cell retrieves the stored documents, metadata, and embeddings from the Chroma collection so we can inspect what has been stored.

```python
# View stored documents

data = vector_store.get(
    include=["embeddings", "documents", "metadatas"]
)

print(data)
```

```python
# view documents
vector_store.get(include=['embeddings','documents', 'metadatas'])
```

**`Cell 10: Similarity Search`**

This cell searches the vector store for the documents that are semantically most similar to the query. k=2 means that the two most similar documents will be returned.

```python
# Search documents

results = vector_store.similarity_search(
    query="Who among these are a bowler?",
    k=2
)

for doc in results:
    print(doc.page_content)
    print(doc.metadata)
    print("-" * 50)
```

**`Cell 11: Similarity Search With Score`**

This cell performs similarity search and also returns a score representing the distance between the query and each document.

```python
# Search with similarity score

results = vector_store.similarity_search_with_score(
    query="Who among these are a bowler?",
    k=2
)

for doc, score in results:
    print("Score:", score)
    print("Content:", doc.page_content)
    print("Metadata:", doc.metadata)
    print("-" * 50)
```

**`Cell 12: Metadata Filtering`**

This cell searches for documents similar to the query but restricts the results to documents whose metadata matches the specified team. This demonstrates how vector similarity and metadata filtering can be combined.

```python
# Metadata filtering

results = vector_store.similarity_search(
    query="Tell me about cricket players",
    k=2,
    filter={"team": "Chennai Super Kings"}
)

for doc in results:
    print(doc.page_content)
    print(doc.metadata)
    print("-" * 50)
```

```python
virat_data = vector_store.get(
    where={"team": "Royal Challengers Bangalore"}
)

print(virat_data["ids"])
```

**`Cell 13: Update a Document`**

This cell replaces the existing Virat Kohli document with updated content while keeping the document in the vector store.

```python
updated_doc1 = Document(
    page_content="Virat Kohli, the former captain of Royal Challengers Bangalore (RCB), is renowned for his aggressive leadership and consistent batting performances. He holds the record for the most runs in IPL history, including multiple centuries in a single season. Despite RCB not winning an IPL title under his captaincy, Kohli's passion and fitness set a benchmark for the league. His ability to chase targets and anchor innings has made him one of the most dependable players in T20 cricket.",
    metadata={"team": "Royal Challengers Bangalore"}
)

vector_store.update_document(
    document_id='d080b654-06c2-40be-b4ba-831e8d1b89b2',
    document=updated_doc1
)

print("Document updated successfully!")
```

```python
# view documents
vector_store.get(include=['embeddings','documents', 'metadatas'])
```

```python
# Delete document

vector_store.delete(ids=['d080b654-06c2-40be-b4ba-831e8d1b89b2'])

print("Document deleted successfully!")
```
